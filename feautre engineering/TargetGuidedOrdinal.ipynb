# Target Guided Ordinal Encoding

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    'city': ['Paris', 'New York', 'Los Angeles', 'Chicago', 'Houston', 'New York'],
    'price': [100, 200, 150, 120, 180, 160]
})
df
```

```python
mean_price = df.groupby('city')['price'].mean()
```

```python
df['city_encoded'] = df['city'].map(mean_price)
df
```

```python
df[['price', 'city_encoded']]
```

```python
import seaborn as sns
df = sns.load_dataset('tips')
```

```python
mean_bill = df.groupby('time')['total_bill'].mean()
```

```python
df['time_encoded'] = df['time'].map(mean_bill)
```

```python
df
```

```python
df[['total_bill','time_encoded']]
```
