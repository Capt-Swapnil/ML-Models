```python
## Miniumum, Maximum, Median, Q1,Q3,IQR
```

```python
import numpy as np
```

```python
lst_marks = [45,78,89,46,45,89,98,67,56,99,96,86,76,65]
minimum,Q1,median,Q3,maximum = np.quantile(lst_marks,[0,0.25,0.5,0.75,1])
```

```python
minimum,Q1,median,Q3,maximum
```

```python
IQR = Q3-Q1
print("IQR:",IQR)
```

```python
lower_fence= Q1-1.5*IQR
higher_fence= Q3+1.5*IQR
```

```python
lower_fence
```

```python
higher_fence
```
